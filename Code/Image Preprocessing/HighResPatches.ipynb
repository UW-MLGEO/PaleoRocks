The goal of this code is to create patches from pdf files that maintain the resolution of the thin sections so that fossils at the smallest scale can still be identified

```python
import random
import pymupdf
from PIL import Image
import os
import io
import pathlib

def PDFrandompatches(filepath, patch_width, patch_height, num_patches):

    # open pdf image
    doc = pymupdf.open(filepath)
    page = doc[0]
    
    # render pdf image to high-resolution pixmap
    pix = page.get_pixmap(matrix=pymupdf.Matrix(matrix_scale, matrix_scale))
    img = Image.open(io.BytesIO(pix.tobytes()))
    
    width, height = img.size
    patches = []
    
    # generate random patches
    for i in range(num_patches):
        # random upper-left corner coordinates
        x1 = random.randint(0, width - patch_width)
        y1 = random.randint(0, height - patch_height)
        x2 = x1 + patch_width
        y2 = y1 + patch_height
        
        print(x1,y1)

        # Crop and save the patch
        patch = img.crop((x1, y1, x2, y2))
        base_name = os.path.splitext(filename)[0]
        patched_name = f"{base_name}_patch_{i+1}.png"
        save_path = os.path.join(output_folder, patched_name)
        patch.save(save_path)
        patches.append(patch)
        
    print(f"Generated {num_patches} patches.")
    return patches
```
